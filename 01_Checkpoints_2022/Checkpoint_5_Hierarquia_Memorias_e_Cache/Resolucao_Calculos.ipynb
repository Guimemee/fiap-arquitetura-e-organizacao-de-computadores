# Checkpoint 5: Análise de Desempenho da Hierarquia de Memória (AMAT)
Cálculo de latência e sensibilidade paramétrica do tempo médio de acesso.


```python
def calc_amat(h1, h2, t1=1.2, t2=8.0, t_ram=65.0):
    m1 = 1.0 - h1
    m2 = 1.0 - h2
    penalty_l1 = t2 + m2 * t_ram
    amat = t1 + m1 * penalty_l1
    return amat

amat_padrao = calc_amat(0.94, 0.85)
amat_otim = calc_amat(0.98, 0.85)
speedup = amat_padrao / amat_otim

print(f'AMAT Padrão (h1=94%):   {amat_padrao:.3f} ns')
print(f'AMAT Otimizado (h1=98%): {amat_otim:.3f} ns')
print(f'Speedup Global:          {speedup:.3f}x')

```

```python
import matplotlib.pyplot as plt
import numpy as np

h1_range = np.linspace(0.85, 0.99, 100)
amat_vals = [calc_amat(h, 0.85) for h in h1_range]

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(h1_range * 100, amat_vals, 'b-', lw=2.5, label='AMAT (ns)')
ax.scatter([94], [amat_padrao], color='orange', s=90, zorder=5, label=f'Original: h1=94% (AMAT={amat_padrao:.2f} ns)')
ax.scatter([98], [amat_otim], color='green', s=90, zorder=5, label=f'Otimizado: h1=98% (AMAT={amat_otim:.2f} ns)')

ax.set_xlabel('Taxa de Acerto da Cache L1 (%)', fontweight='bold')
ax.set_ylabel('Tempo Médio de Acesso AMAT (ns)', fontweight='bold')
ax.set_title('Impacto da Eficiência da Cache L1 no AMAT do Processador', fontweight='bold')
ax.legend(loc='upper right')
ax.grid(True, alpha=0.5)
plt.tight_layout()
plt.show()

```
