# Checkpoint 1: Sistemas de Numeração e Conversões de Base
Cálculo de conversões canônicas em Python para validação estrutural.


```python
val_dec = 357
print(f'Decimal: {val_dec}')
print(f'Binario: {bin(val_dec)[2:]}')
print(f'Octal:   {oct(val_dec)[2:]}')
print(f'Hex:     {hex(val_dec)[2:].upper()}')

# Conversao binaria fracionaria
b_frac = '1011'
dec_frac = sum(int(bit) * (2 ** -(i + 1)) for i, bit in enumerate(b_frac))
print(f'Fração binária 0.{b_frac} em decimal: {dec_frac}')

```

```python
import matplotlib.pyplot as plt
import numpy as np

bases = ['Binário\n(101100101)', 'Octal\n(545)', 'Decimal\n(357)', 'Hexadecimal\n(165)']
digitos = [9, 3, 3, 3]

fig, ax = plt.subplots(figsize=(7, 4))
bars = ax.bar(bases, digitos, color=['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728'], width=0.5)
ax.set_ylabel('Número de Dígitos para Representar 357', fontweight='bold')
ax.set_title('Densidade da Informação por Base Numérica', fontweight='bold')
for bar in bars:
    yval = bar.get_height()
    ax.text(bar.get_x() + bar.get_width()/2.0, yval + 0.2, int(yval), ha='center', va='bottom', fontweight='bold')
ax.set_ylim(0, 11)
plt.tight_layout()
plt.show()

```
