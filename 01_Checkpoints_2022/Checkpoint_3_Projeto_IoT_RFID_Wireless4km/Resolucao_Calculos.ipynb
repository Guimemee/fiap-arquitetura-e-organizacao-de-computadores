# Checkpoint 3: Link Budget de Comunicação Sem Fio IoT (4 km)
Cálculo de propagação de RF e dimensionamento de enlace LoRa/RFID.


```python
import numpy as np

d = 4.0      # km
f = 915.0    # MHz
P_tx = 14.0  # dBm
G_tx = 2.0   # dBi
G_rx = 6.0   # dBi
S_rx = -120.0 # dBm

fspl = 20 * np.log10(d) + 20 * np.log10(f) + 32.44
p_rx = P_tx + G_tx + G_rx - fspl
margem = p_rx - S_rx

print(f'Perda no Espaço Livre (FSPL): {fspl:.2f} dB')
print(f'Potência Recebida (P_rx):      {p_rx:.2f} dBm')
print(f'Sensibilidade Receptor:        {S_rx:.2f} dBm')
print(f'Margem de Desvanecimento:      {margem:.2f} dB (Viável: {margem > 20})')

```

```python
import matplotlib.pyplot as plt
import numpy as np

distancias = np.linspace(0.5, 10.0, 200)
perdas = 20 * np.log10(distancias) + 20 * np.log10(915) + 32.44
pot_rec = 14 + 2 + 6 - perdas

fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(distancias, pot_rec, 'b-', lw=2.5, label='Potência Recebida Prx (dBm)')
ax.axhline(-120, color='r', linestyle='--', label='Sensibilidade Limite (-120 dBm)')
ax.scatter([4.0], [-81.71], color='darkgreen', s=90, zorder=5, label='Ponto Operacional (4 km, -81.7 dBm)')

ax.set_xlabel('Distância de Propagação (km)', fontweight='bold')
ax.set_ylabel('Potência de Sinal (dBm)', fontweight='bold')
ax.set_title('Atenuação de Potência vs Distância em 915 MHz (Enlace LoRa)', fontweight='bold')
ax.legend(loc='upper right')
ax.grid(True, alpha=0.5)
plt.tight_layout()
plt.show()

```
