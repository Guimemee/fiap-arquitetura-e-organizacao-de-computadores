# Checkpoint 6: Dimensionamento de Barramentos PCIe e Controladores DMA
Cálculo de Throughput e tempos de transferência de barramentos de E/S.


```python
gt_s = 8.0 # GT/s por pista PCIe 3.0
encoding_eff = 128.0 / 130.0
lanes = 4

bitrate_net_gbps = gt_s * encoding_eff * lanes
throughput_gBps = bitrate_net_gbps / 8.0

tamanho_dados_gb = 2.0
t_dma = tamanho_dados_gb / throughput_gBps
t_polling = t_dma * 1.45

print(f'PCIe 3.0 x4 Throughput Efetivo: {throughput_gBps:.3f} GB/s')
print(f'Tempo de transferência DMA:      {t_dma * 1000:.1f} ms')
print(f'Tempo com overhead de CPU:      {t_polling * 1000:.1f} ms')
print(f'Economia de tempo com DMA:      {(t_polling - t_dma) * 1000:.1f} ms')

```

```python
import matplotlib.pyplot as plt

interfaces = ['PCIe 2.0 x4', 'PCIe 3.0 x4', 'PCIe 4.0 x4', 'SATA III', 'USB 3.0']
bandwidths = [2.0, 3.94, 7.88, 0.6, 0.625] # GB/s

fig, ax = plt.subplots(figsize=(8, 4.5))
bars = ax.barh(interfaces, bandwidths, color=['#7f7f7f', '#1f77b4', '#2ca02c', '#d62728', '#ff7f0e'])
ax.set_xlabel('Largura de Banda Efetiva (GB/s)', fontweight='bold')
ax.set_title('Comparativo de Throughput de Barramentos de E/S de Computadores', fontweight='bold')
for bar in bars:
    w = bar.get_width()
    ax.text(w + 0.15, bar.get_y() + bar.get_height()/2.0, f'{w:.2f} GB/s', ha='left', va='center', fontweight='bold')
ax.set_xlim(0, 9.5)
ax.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()

```
