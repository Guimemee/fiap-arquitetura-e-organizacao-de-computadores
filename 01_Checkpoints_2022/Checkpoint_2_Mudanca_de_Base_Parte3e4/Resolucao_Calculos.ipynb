# Checkpoint 2: Complemento de 2 e Ponto Flutuante IEEE-754
Implementação das conversões de Complemento de 2 e IEEE-754 32-bit.


```python
import struct

# Validacao Complemento de 2
a = 85
b = 42
sub = a - b
comp2_b = (~b + 1) & 0xFF
res_bin = (a + comp2_b) & 0xFF
print(f'85 em 8 bits: {a:08b}')
print(f'-42 em Comp2: {comp2_b:08b}')
print(f'Soma (85 - 42): {res_bin:08b} = {res_bin}')

# Validacao IEEE 754
num = -18.625
packed = struct.pack('>f', num)
bits = ''.join(f'{b:08b}' for b in packed)
hex_val = packed.hex().upper()
print(f'\nIEEE 754 (-18.625):')
print(f'Sinal:    {bits[0]}')
print(f'Expoente: {bits[1:9]}')
print(f'Mantissa: {bits[9:]}')
print(f'Hex:      0x{hex_val}')

```

```python
import matplotlib.pyplot as plt

campos = ['Sinal (1b)', 'Expoente (8b)', 'Mantissa (23b)']
tamanhos = [1, 8, 23]
cores = ['#d62728', '#ff7f0e', '#1f77b4']

fig, ax = plt.subplots(figsize=(8, 3))
left = 0
for nome, tam, cor in zip(campos, tamanhos, cores):
    ax.barh(0, tam, left=left, color=cor, edgecolor='black', height=0.5, label=f'{nome}')
    ax.text(left + tam/2, 0, f'{nome}\n{tam} bits', ha='center', va='center', color='white', fontweight='bold', fontsize=10)
    left += tam

ax.set_xlim(0, 32)
ax.set_ylim(-0.5, 0.5)
ax.axis('off')
ax.set_title('Estrutura de 32 Bits do Formato IEEE 754 (Precisão Simples)', fontweight='bold')
plt.tight_layout()
plt.show()

```
